# Precision-ladder demo (the hardware case study)

This notebook walks the precision ladder — FP32, FP16, BF16, FP8, and (on Blackwell only) NVFP4/MXFP4 — and shows how run-to-run determinism changes with precision.

It degrades gracefully: on a CPU or a non-Blackwell GPU the unsupported rungs report a clear skip message and the rest still run. You do **not** need Blackwell hardware to run this notebook; you need it only to observe the FP4 rung.

```python
import detequal.precision as precision

# Which rungs does THIS machine support?
for rung in precision.probe_support():
    status = "supported" if rung.supported else rung.reason
    print(f"{rung.precision.value:10s} {status}")
```

The full per-rung sweep (`precision.run_ladder`) is still under construction — the capability probe above is what is live today. Once the sweep lands, this cell produces the cross-precision summary that feeds the empirical study: fraction of layers flagged and divergence-score distribution per precision.
