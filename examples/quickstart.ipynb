# detequal quickstart

Run a model N times, hook every module, and see which layers diverge across runs. This notebook uses a small deterministic model so it runs anywhere (CPU included) with no divergent layers; swap in your own model to find real nondeterminism.

```python
import torch
import torch.nn as nn

import detequal

model = nn.Sequential(nn.Linear(16, 32), nn.ReLU(), nn.Linear(32, 8))
batch = torch.randn(4, 16)

report = detequal.diagnose(model, batch, n_runs=5)
report.summary()
```

## Comparison mode

Pass `force_deterministic=True` to characterize behavior with `torch.use_deterministic_algorithms(True)` enabled, contrasting it against the default (switch off) case.

```python
report_det = detequal.diagnose(model, batch, n_runs=5, force_deterministic=True)
report_det.summary()
```
